```python
import pandas as pd
import numpy as np

np.random.seed(15)
n_samples = 1500

time_slots = np.random.choice(['Morning', 'Afternoon', 'Evening'], size=n_samples, p=[0.3, 0.5, 0.2])
days = np.random.choice(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'], size=n_samples)
weather = np.random.choice(['Sunny', 'Rainy', 'Overcast'], size=n_samples, p=[0.6, 0.2, 0.2])
zone_types = np.random.choice(['Quiet_Zone', 'Collaborative_Zone'], size=n_samples, p=[0.5, 0.5])
academic_phase = np.random.choice(['Regular_Teaching', 'Assignment_Peak', 'Exam_Period'], size=n_samples, p=[0.5, 0.3, 0.2])
power_outlet = np.random.choice(['High_Demand_Plug', 'Standard_Desk'], size=n_samples, p=[0.6, 0.4])
post_lecture = np.random.choice([True, False], size=n_samples, p=[0.3, 0.7]) # means while.

# 计算拥挤得分
congestion_score = (
    (time_slots == 'Afternoon') * 1.5 + 
    (days == 'Friday') * 1.0 + 
    (weather == 'Rainy') * 1.0 + 
    (academic_phase == 'Exam_Period') * 3.0 + 
    (academic_phase == 'Assignment_Peak') * 1.8 +  
    (zone_types == 'Collaborative_Zone') * 1.2 +   
    (power_outlet == 'High_Demand_Plug') * 1.5 +    
    np.random.normal(0, 0.8, n_samples)
)

# 【核心修改】改用 pd.cut 并手动指定切分边界（Bins）
# 这样就不会强制均分，而是根据真实的得分高低来自然分布类别比例！
bins = [-np.inf, 2.2, 4.5, np.inf]
levels = pd.cut(congestion_score, bins=bins, labels=['Free', 'Moderate', 'Crowded'])

df = pd.DataFrame({
    'Time_Slot': time_slots,
    'Day_Of_Week': days,
    'Weather': weather,
    'Academic_Phase': academic_phase,
    'Zone_Type': zone_types,
    'Power_Outlet': power_outlet,
    'Post_Lecture': post_lecture,
    'Crowding_Level': levels
})

print("=== 修复后的自然分布数据加载成功 ===")
print(f"总样本量: {len(df)}")
print("\n各拥挤等级样本百分比分布 (Class Distribution):")
print(df['Crowding_Level'].value_counts(normalize=True) * 100)

display(df.head())
```

```python
df.to_csv('study_space_data.csv', index=False)
```
